```python
import numpy as np 
import matplotlib.pyplot as plt 
```

```python
x = np.array([0.5, 1.0, 1.5, 2.0]) 
y = np.array([2.0, 3.0, 4.0, 5.0]) 
 
params0 = np.array([0.5, 0.5, 1.0, 0.0]) 
```

```python
def forward(params, x): 
 
    w1, b1, w2, b2 = params
 
    z1 = w1 * x + b1
    h = np.maximum(z1,0)
    yhat = w2 * h + b2
 
    return z1, h, yhat 
 
 
def batch_loss(params, x, y): 
 
    z1, h, yhat = forward(params, x) 
 
    assert yhat.shape == y.shape 
 
    return np.mean((yhat - y)**2)

z1, h, yhat = forward(params0, x)
loss = batch_loss(params0, x, y)
sq_error = (yhat - y) ** 2

print("z1: ", z1)
print("h: ", h)
print("yhat: ", yhat)
print("Loss: ", loss)
```

```python
def batch_gradients(params, x, y): 
 
    w1, b1, w2, b2 = params 
 
    z1, h, yhat = forward(params, x) 
 
    N = len(x) 
 
    g_yhat = (2/N)*(yhat - y)
    g_w2 = np.sum(g_yhat * h)
    g_b2 = np.sum(g_yhat)
    g_h = g_yhat * w2
    g_z1 = g_h * (z1 > 0).astype(float)
    g_w1 = np.sum(g_z1 * x)
    g_b1 = np.sum(g_z1)
 
    return np.array([g_w1, g_b1, g_w2, g_b2]) 

grads0 = batch_gradients(params0, x, y)
print("Gradients:", grads0)
```

1. Where does the factor 1/N enter the gradients, and why?
<br>It enters from the MSE loss function because loss uses an average. When we calculate gradients, this 1/N scaling factor naturally stays to average the error across all N examples.

2. Why do the shared-parameter gradients require a sum over the examples?
<br>Because the parameters w and b are shared and used by all examples in the batch. Since every example contributes to the total error, we must sum up all individual contributions to get the total parameter gradient.

4. What is the difference between the per-example vectors of shape (4,) and the returned parameter-gradient vector of shape (4,)?
<br>A per-example vector holds intermediate values for each of the 4 data points. The parameter-gradient vector holds the single total gradient for each of the 4 parameters.

5. Why is the output of this model not a probability?
<br>Because this is a regression model with a linear output layer and no Sigmoid activation. Its predictions are continuous numbers, not bounded values between 0 and 1.

```python
def numerical_gradient(fn, params, eps=1e-5):
    gradient = np.zeros_like(params, dtype=float)

    for j in range(len(params)):
        plus = params.copy()
        minus = params.copy()

        plus[j] += eps
        minus[j] -= eps

        gradient[j] = (fn(plus) - fn(minus)) / (2 * eps)

    return gradient


analytic = batch_gradients(params0, x, y)

numeric = numerical_gradient(
    lambda p: batch_loss(p, x, y),
    params0
)

print("analytic:", analytic)
print("numeric :", numeric)

print("max absolute difference:", np.max(np.abs(analytic - numeric)))

assert np.max(np.abs(analytic - numeric)) < 1e-6

print("Task 3A OK")
```

```python
params = params0.copy() 
eta = 0.01 
 
loss_history = [batch_loss(params, x, y)] 
 
for step in range(200): 
 
    gradients = batch_gradients(params, x, y)
    params = params - eta * gradients
 
    loss_history.append(batch_loss(params, x, y)) 
 
print("final params:", params) 
print("final loss  :", loss_history[-1]) 
 
assert len(loss_history) == 201 
assert loss_history[-1] < loss_history[0] 
 
print("Task 3B OK") 
```
