```python
import numpy as np
```

```python
x = np.array([1.0, 2.0, 3.0])
y = np.array([2.0, 4.0, 5.0])

w = 0.0
b = 0.0
lr = 0.1 
y_V = w * x + b
```

```python
initial_loss = np.mean((y_V - y) ** 2)
print("Initial Loss:", initial_loss)
```

```python
error = y_V - y

N = len(y)
dw = (2 / N) * np.sum(error * x)
db = (2 / N) * np.sum(error)

print("Gradients dw, db =",dw,db)
```

```python
w_new = w - lr * dw
b_new = b - lr * db

print("Updated parameters w, b=",w_new, b_new)
```

```python
y_V_new = w_new * x + b_new
new_loss = np.mean((y_V_new - y) ** 2)

print("New Loss:", new_loss)
print("Loss decreased =", new_loss < initial_loss)
```

1. Why are both gradients negative?
<p>Because all predictions y_V=0 were smaller than the target values y, making the residuals (y_V-y) negative.</p>

2. Why does subtracting a negative gradient increase w and b?
<p>Because two negatives make a positive</p>

3. Why must both gradients be calculated before updating the parameters?
<p>If we change w first, we will use the wrong w value when calculating the gradient for b.</p>

4. The model will do gradient ascent. The loss will go up instead of down, making predictions worse.

5. Why is checking whether the loss decreased a useful debugging tool?
<p>It is an easy check: if loss goes down, the code works; if loss goes up, there is a bug in the code or signs.</p>
