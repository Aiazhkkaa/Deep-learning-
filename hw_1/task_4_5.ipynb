```python
import numpy as np 
```

```python
X = np.array([ 
    [1.0, 2.0], 
    [2.0, 0.5], 
    [3.0, 1.0], 
    [4.0, 3.0], 
]) 
 
w = np.array([2.0, -1.0])   # shape (2,) 
b = 0.5 
 
 
def predict(X, w, b): 
   return (X @ w) + b
 
yhat = predict(X, w, b) 
print("yhat shape:", yhat.shape) 
print("yhat      :", yhat) 
 
# --- Now deliberately introduce a shape bug --- 
w_bad = w.reshape(2, 1)   # shape (2, 1) instead of (2,) 
 
yhat_bad = X @ w_bad + b 
print("yhat_bad shape:", yhat_bad.shape) 
print("yhat_bad      :", yhat_bad) 
 
assert yhat.shape == (4,) 
assert yhat_bad.shape == (4, 1) 
 
print("Task 4 OK") 
```

#### 1. What does each dimension of X represent?
<p>Dimension 0 (4) represents the number of data samples (examples).</p>
<p>Dimension 1 (2) represents the number of input features per sample.</p>

#### 2. What does the shape (2,) of w represent?
<p>It is a 1D vector containing one weight parameter for each of the two features.</p>

#### 3. Why is the correct prediction shape (4,)?
<p>The dataset contains 4 samples, so the model must output one scalar prediction per sample.</p>

#### 4. Why does changing w to shape (2,1) change the prediction shape?
<p>Multiplying a 2D matrix of shape (4, 2) by a 2D column vector of shape (2, 1) yields a 2D matrix of shape (4, 1).</p>

#### 5. Why can a shape such as (4,1) be dangerous even though the values look correct?
<p>Combining a (4, 1) array with a 1D target array (4,) triggers NumPy's broadcasting rule, creating a (4, 4) matrix instead of performing element-wise operations.</p>

```python
def mse_loop(y,y_pred):
    total_skerror=0.0

    for i in range(len(y)):
       
       total_skerror+=((y_pred[i]-y[i])**2)

    return total_skerror/len(y)   
```

```python
def mse_numpy(y,y_pred):
    return np.mean((y_pred-y)**2)
```

```python
print("MSE(python loop):" , mse_loop(y,y_pred)) 
print("MSE(numpy):" , mse_numpy(y,y_pred)) 
```
